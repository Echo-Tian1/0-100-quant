一句话版：

> 因子投资，就是把“股票为什么涨跌”这件事拆成一组可解释、可检验、可组合的规律。

## 1. 先建立直觉：股票收益像一杯混合饮料

假设你买了一只股票，20 天后涨了 8%。你可能会说：“我选对了！”

但因子研究员会先问：

- 这 20 天大盘是不是也涨了？
- 小盘股是不是整体表现更好？
- 低估值股票是不是整体反弹？
- 高盈利质量公司是不是被市场奖励？
- 这只股票是不是刚好暴露在这些共同规律上？

也就是说，一只股票的收益不是一个单一来源，而像一杯混合饮料：


股票收益 = 市场整体收益 + 风格/因子收益 + 公司特有收益 + 噪声


再用投资语言说：


股票收益 = Beta 收益 + 因子暴露收益 + Alpha + 随机扰动


这里的重点不是公式，而是思维方式：

> 不要急着说“我赚的钱来自能力”。先拆解：它来自市场、来自风格、来自真正的选股，还是来自运气。

## 2. 三个核心词：Alpha、Beta、Factor

### 2.1 Beta：你跟着系统性风险一起摆动的程度

最简单的 Beta 是市场 Beta。

如果市场涨 1%，某股票平均涨 1.2%，这只股票的市场 Beta 大约是 1.2。

直觉上：

- Beta = 1：和市场差不多同频
- Beta > 1：比市场更敏感，进攻性更强
- Beta < 1：比市场更稳
- Beta < 0：和市场反向，但现实中普通股票很少长期如此

经典的单因子模型：


R_i - R_f = alpha_i + beta_i * (R_m - R_f) + error_i


其中：

- `R_i`：股票 i 的收益率
- `R_f`：无风险收益率
- `R_m`：市场收益率
- `R_m - R_f`：市场风险溢价
- `beta_i`：股票对市场的暴露
- `alpha_i`：模型解释不了的平均超额收益
- `error_i`：短期噪声

为了初学清晰，今天的 Python 示例先忽略无风险收益率，写成：


stock_return = alpha + beta * market_return + noise


### 2.2 Alpha：模型解释不掉的那部分

Alpha 很容易被误解成“我很强，所以多赚的钱”。

更准确地说：

> Alpha 是在某个风险模型下，扣掉已知风险暴露后，剩下的平均收益。

注意这句话里最关键的是“在某个风险模型下”。

假设你买了一堆小盘低估值股票，过去一年跑赢大盘。你以为这是 Alpha，但如果加入“小盘因子”和“价值因子”后，这些超额收益都能被解释掉，那它可能不是 Alpha，而是因子暴露。

所以：

- 在 CAPM 模型下看起来是 Alpha 的东西，
- 到了 Fama-French 三因子模型里可能变成 SMB 或 HML 暴露，
- 到了更复杂的风格模型里可能又被拆得更细。

Alpha 不是绝对概念，它依赖你使用的基准模型。

### 2.3 Factor：可解释、可度量、可重复验证的共同规律

一个因子通常要满足三件事：

1. **可解释**：背后有经济、行为、风险补偿或市场结构逻辑。
2. **可度量**：能用数据算出来，比如 PB、ROE、过去 60 日收益率、波动率、换手率。
3. **可验证**：能在历史数据里检验它是否和未来收益、风险或组合表现有关。

例子：

| 因子类别 | 常见指标 | 直觉 |
| --- | --- | --- |
| 价值 | PE、PB、EP、BP、股息率 | 买得便宜是否有补偿 |
| 质量 | ROE、ROA、毛利率、现金流质量 | 好公司是否长期更稳 |
| 动量 | 20 日、60 日、120 日收益 | 强者是否继续强 |
| 波动率 | 历史波动率、特质波动率 | 高波动是否被错误定价 |
| 流动性 | 换手率、成交额、Amihud | 难交易资产是否有补偿 |
| 规模 | 总市值、流通市值 | 小公司是否有风险溢价 |

一句很实用的判断：

> 指标只是原材料。能形成稳定预测、风险解释或组合收益差异时，它才开始像一个因子。

## 3. 用 Python 亲手看见 Alpha 和 Beta

这一节不需要真实行情数据。我们先造一个简化世界：市场每天有收益，几只股票围绕市场上下波动，每只股票有不同的 Beta 和 Alpha。

你可以把下面代码复制到 Jupyter Notebook、VS Code、PyCharm 或普通 `.py` 文件中运行。

### 3.1 准备环境


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (10, 5)
plt.rcParams["axes.grid"] = True

rng = np.random.default_rng(42)
```

### 3.2 模拟市场收益


```python
dates = pd.bdate_range("2024-01-01", periods=252)

# 假设市场日均收益约 0.04%，日波动约 1%
market_ret = pd.Series(
    rng.normal(loc=0.0004, scale=0.01, size=len(dates)),
    index=dates,
    name="market_ret"
)

market_ret.head()
```

这里的 `252` 代表一年大约 252 个交易日。

### 3.3 模拟 4 只股票


```python
stock_settings = {
    "稳健低波": {"alpha": 0.0001, "beta": 0.6, "noise": 0.008},
    "市场同步": {"alpha": 0.0000, "beta": 1.0, "noise": 0.010},
    "进攻成长": {"alpha": 0.0002, "beta": 1.4, "noise": 0.014},
    "伪Alpha": {"alpha": 0.0000, "beta": 1.8, "noise": 0.012},
}

stock_ret = pd.DataFrame(index=dates)

for name, cfg in stock_settings.items():
    stock_ret[name] = (
        cfg["alpha"]
        + cfg["beta"] * market_ret
        + rng.normal(0, cfg["noise"], len(dates))
    )

stock_ret.head()
```

观察名字：

- `稳健低波`：低 Beta，市场大涨时可能没那么猛，市场大跌时也可能更稳。
- `市场同步`：Beta 接近 1。
- `进攻成长`：高 Beta，外加一点真实 Alpha。
- `伪Alpha`：看起来很猛，但主要是因为 Beta 很高。

### 3.4 看累计收益


```python
wealth = (1 + pd.concat([market_ret, stock_ret], axis=1)).cumprod()
wealth.columns = ["市场"] + list(stock_ret.columns)

wealth.plot(title="模拟市场与股票累计净值")
plt.ylabel("累计净值")
plt.show()
```

你可能会看到 `伪Alpha` 在牛市阶段很漂亮。但问题是：它到底是有 Alpha，还是只是把市场 Beta 加到了 1.8？

这就是因子研究的价值：把漂亮的收益曲线拆开看。

### 3.5 用最小二乘法估计 Alpha 和 Beta

我们对每只股票做一个简单回归：


stock_return = alpha + beta * market_return + noise


不用额外安装 `statsmodels`，用 NumPy 就可以做。


```python
def estimate_alpha_beta(y, x):
    """
    y: 股票收益率序列
    x: 市场收益率序列
    返回：日 alpha、市场 beta
    """
    X = np.column_stack([np.ones(len(x)), x.values])
    coef = np.linalg.lstsq(X, y.values, rcond=None)[0]
    alpha_daily, beta = coef
    return alpha_daily, beta

rows = []

for name in stock_ret.columns:
    alpha_daily, beta = estimate_alpha_beta(stock_ret[name], market_ret)
    alpha_annual = (1 + alpha_daily) ** 252 - 1
    rows.append({
        "股票": name,
        "估计日Alpha": alpha_daily,
        "估计年化Alpha": alpha_annual,
        "估计Beta": beta,
        "年化波动率": stock_ret[name].std() * np.sqrt(252),
        "年化收益率": (1 + stock_ret[name]).prod() ** (252 / len(stock_ret)) - 1,
    })

result = pd.DataFrame(rows)
result
```

你会发现一个很重要的现象：

> 收益率高，不等于 Alpha 高。很多收益只是来自更高的 Beta。

### 3.6 画出某只股票的 Beta


```python
name = "伪Alpha"
alpha_daily, beta = estimate_alpha_beta(stock_ret[name], market_ret)

x_line = np.linspace(market_ret.min(), market_ret.max(), 100)
y_line = alpha_daily + beta * x_line

plt.scatter(market_ret, stock_ret[name], alpha=0.5, label="每日收益点")
plt.plot(x_line, y_line, color="red", label=f"回归线：alpha={alpha_daily:.5f}, beta={beta:.2f}")
plt.axhline(0, color="black", linewidth=1)
plt.axvline(0, color="black", linewidth=1)
plt.title(f"{name}：市场收益 vs 股票收益")
plt.xlabel("市场日收益")
plt.ylabel("股票日收益")
plt.legend()
plt.show()
```

这张图里：

- 每个点是一日收益。
- 横轴是市场收益。
- 纵轴是股票收益。
- 红线斜率就是 Beta。
- 红线截距就是日 Alpha。

如果斜率很陡，这只股票可能不是“神奇”，只是更依赖市场方向。

## 4. 从 Beta 到多因子：收益可以被更多共同规律解释

单因子 CAPM 只问一个问题：

> 这只股票和市场的关系是什么？

但股票收益显然不只由市场决定。比如：

- 小盘股和大盘股经常有不同表现。
- 低估值股票和高估值股票表现可能不同。
- 高盈利质量公司和低盈利质量公司表现可能不同。
- 高动量股票和低动量股票表现可能不同。

于是我们把模型扩展：


股票收益
= alpha
+ beta_market * 市场因子
+ beta_value  * 价值因子
+ beta_size   * 规模因子
+ beta_quality * 质量因子
+ beta_momentum * 动量因子
+ noise


这里每个 `beta_xxx` 都表示暴露：

- 对市场因子暴露高：市场一动，它更容易跟着动。
- 对价值因子暴露高：价值风格强时，它更容易受益。
- 对动量因子暴露高：趋势风格强时，它更容易受益。

这就是多因子框架的核心：

> 先定义共同规律，再测量资产对这些规律的暴露，最后分析收益来自哪里。

---

## 5. Fama-French 简介：为什么经典？

Fama-French 模型是因子投资绕不开的入口。

### 5.1 从 CAPM 到三因子

CAPM 认为市场风险可以解释资产收益差异：


超额收益 = alpha + beta_market * 市场超额收益


但实证中，人们发现仅靠市场 Beta 解释不了很多现象。比如：

- 小市值股票长期表现和大市值股票不同。
- 低估值股票长期表现和高估值股票不同。

Fama 和 French 在经典三因子模型中加入了两个因子：

| 因子 | 名称 | 含义 |
| --- | --- | --- |
| MKT | Market | 市场风险溢价 |
| SMB | Small Minus Big | 小市值减大市值，代表规模因子 |
| HML | High Minus Low | 高账面市值比减低账面市值比，代表价值因子 |

模型变成：


R_i - R_f
= alpha_i
+ beta_mkt * MKT
+ beta_smb * SMB
+ beta_hml * HML
+ error_i


### 5.2 五因子模型

后来 Fama-French 又扩展到五因子，加入：

| 因子 | 含义 |
| --- | --- |
| RMW | Robust Minus Weak，高盈利质量减低盈利质量 |
| CMA | Conservative Minus Aggressive，保守投资减激进投资 |

五因子模型的大意是：

> 股票收益不仅和市场、规模、价值有关，也和盈利能力、投资风格有关。

### 5.3 你今天只需要抓住这件事

Fama-French 最重要的启发不是“记住几个英文缩写”，而是：

> 超额收益可以被系统性风格拆解。你以为的 Alpha，可能只是没有被识别出来的因子暴露。

## 6. 小实验：从股票指标构造一个“价值因子”

下面我们模拟一个截面数据。每一行是一只股票，包含：

- `pb`：市净率，越低通常越偏价值。
- `roe`：净资产收益率，越高通常质量越好。
- `future_20d_ret`：未来 20 日收益率，这在真实研究中是明天要学的“标签”。

今天我们先不深究标签构造，只用它帮助理解因子。

### 6.1 构造模拟股票池


```python
n = 300

universe = pd.DataFrame({
    "ticker": [f"Stock_{i:03d}" for i in range(n)],
    "market_cap": rng.lognormal(mean=10, sigma=1.0, size=n),
    "pb": rng.lognormal(mean=0.3, sigma=0.6, size=n),
    "roe": rng.normal(loc=0.10, scale=0.06, size=n),
})

universe.head()
```

### 6.2 把 PB 变成价值暴露

PB 越低越“便宜”，但数值越低越好，不太方便直接当因子分数。我们可以把它转成：


value_score = -标准化(log(PB))


这样：

- PB 越低，`log(PB)` 越低。
- 加一个负号后，价值分数越高。


```python
def zscore(s):
    return (s - s.mean()) / s.std()

universe["value_score"] = -zscore(np.log(universe["pb"]))
universe["quality_score"] = zscore(universe["roe"])

universe[["ticker", "pb", "value_score", "roe", "quality_score"]].head()
```

### 6.3 模拟未来收益：让价值因子有一点有效性

真实世界里我们不会提前知道未来收益。这里为了学习，故意让未来收益和价值分数有一点关系。


```python
noise = rng.normal(0, 0.06, size=n)

universe["future_20d_ret"] = (
    0.01 * universe["value_score"]
    + 0.006 * universe["quality_score"]
    + noise
)

universe[["ticker", "value_score", "quality_score", "future_20d_ret"]].head()
```

### 6.4 做一个最简单的分组检验

我们按 `value_score` 从低到高分成 5 组，看每组未来收益均值。


```python
universe["value_group"] = pd.qcut(
    universe["value_score"],
    q=5,
    labels=["G1 最低价值", "G2", "G3", "G4", "G5 最高价值"]
)

group_ret = universe.groupby("value_group", observed=True)["future_20d_ret"].mean()
group_ret
```

画图：


```python
group_ret.plot(kind="bar", title="价值因子分组后的未来 20 日平均收益")
plt.ylabel("未来 20 日收益")
plt.xticks(rotation=30)
plt.show()
```

如果 `G5 最高价值` 的平均收益高于 `G1 最低价值`，说明这个模拟世界里价值因子有一定预测力。

### 6.5 做一个极简多空组合


```python
long_ret = universe.loc[universe["value_group"] == "G5 最高价值", "future_20d_ret"].mean()
short_ret = universe.loc[universe["value_group"] == "G1 最低价值", "future_20d_ret"].mean()
long_short_ret = long_ret - short_ret

print(f"做多高价值组收益: {long_ret:.2%}")
print(f"做空低价值组收益: {short_ret:.2%}")
print(f"价值因子多空收益: {long_short_ret:.2%}")
```

这就是最朴素的因子检验思想：

> 如果一个因子真的有用，那么高分组和低分组在未来收益上应该有可观察的差异。

- 第 3 天：计算 IC

## 8. 一套完整的因子研究流程

你可以把因子研究理解成一条流水线。


想法
  ↓
因子定义
  ↓
数据准备
  ↓
清洗与预处理
  ↓
有效性检验
  ↓
组合回测
  ↓
归因与风险分析
  ↓
上线或放弃


### 8.1 想法

例子：

> 低 PB 股票是否在未来 20 天有更高收益？

好的因子想法通常来自：

- 金融理论：风险补偿、资本结构、盈利能力。
- 行为金融：过度反应、反应不足、锚定、追涨杀跌。
- 市场结构：流动性限制、交易成本、机构偏好、指数调仓。
- 数据观察：某类指标长期和收益或风险有关。

### 8.2 因子定义

把想法变成公式：


价值因子 = -zscore(log(PB))


注意：因子定义要具体到别人可以复现。

不要写：


选择便宜股票


要写：


每个调仓日，取最近可得 PB，计算 -zscore(log(PB))，值越大代表越便宜。


### 8.3 数据准备

常见数据：

- 行情：开盘价、收盘价、复权价、成交量、成交额。
- 财务：利润表、资产负债表、现金流量表。
- 估值：PE、PB、PS、EV/EBITDA。
- 行业：申万行业、中信行业、GICS。
- 风险：市值、Beta、波动率、流动性。

### 8.4 清洗与预处理

后面第 14-15 天会重点学。

今天先知道几个名字：

- 去极值：防止少数极端值把结果带偏。
- 标准化：把不同量纲变成可比较的分数。
- 行业中性化：避免因子只是行业差异。
- 市值中性化：避免因子只是大盘小盘差异。

### 8.5 有效性检验

常见检验：

- IC：因子值和未来收益的相关性。
- Rank IC：因子排名和未来收益排名的相关性。
- ICIR：IC 的稳定性。
- 分层回测：高分组是否持续跑赢低分组。
- 多空组合：做多高分组、做空低分组的收益。

### 8.6 组合构建

当因子有效后，还要思考：

- 怎么给股票分配权重？
- 单因子还是多因子？
- 是否行业中性？
- 是否控制换手率？
- 是否考虑交易成本？
- 是否限制个股权重、行业权重、风格暴露？

### 8.7 归因与风险分析

策略赚钱后，还要问：

- 赚钱来自哪个因子？
- 是来自市场上涨，还是来自真正的选股？
- 最近失效是因为风格切换，还是因子本身坏了？
- 回撤来自行业集中、市值暴露、流动性冲击，还是个股事件？

## 9. 初学者最容易踩的 8 个坑

### 坑 1：把高收益直接当 Alpha

如果一个策略只是高 Beta，它在牛市中会显得很聪明，熊市中会暴露本质。

正确做法：

> 先和市场、行业、市值、常见风格因子比较，再谈 Alpha。

### 坑 2：把财务指标直接当因子

PB 是指标，`-zscore(log(PB))` 才更像一个可用于研究的价值因子版本。

因子需要明确：

- 方向：越大越好还是越小越好？
- 频率：每天、每周、每月更新？
- 数据时点：是否避免未来函数？
- 处理方式：是否去极值、标准化、中性化？

### 坑 3：忘记未来函数

比如你在 2024 年 1 月 1 日使用了 2024 年 3 月才披露的财务数据，这就是未来函数。

因子研究里，时间点比公式本身更重要。

### 坑 4：只看一次回测结果

一次漂亮回测可能来自偶然、参数过拟合、市场阶段适配。

更可靠的检查：

- 分年度表现
- 滚动 IC
- 不同股票池
- 不同持有期
- 不同调仓频率
- 加入交易成本

### 坑 5：忽略交易成本

高换手因子看起来很美，实盘可能被手续费、冲击成本和滑点吃掉。

### 坑 6：把相关性当因果

某因子和收益相关，不代表它导致收益。

你至少要问：

- 有没有经济逻辑？
- 有没有稳定性？
- 有没有在不同市场阶段保持方向？
- 是否只是另一个已知因子的影子？

### 坑 7：把因子收益当永久规律

因子会拥挤、会衰减、会阶段性失效。

一个成熟研究员会问：

> 这个因子什么时候赚钱？什么时候亏钱？为什么？

### 坑 8：忽略风险暴露

一个策略看似是价值策略，实际可能同时重仓金融、低成长、小市值、高杠杆。

不做归因，就不知道自己到底买了什么。

## 11. 自测题

### 题 1

某策略一年收益 30%，同期市场收益 25%。它一定有 Alpha 吗？

答案：不一定。它可能只是比市场 Beta 更高。如果策略 Beta 是 1.3，那么市场涨 25% 时，策略本来就可能涨得更多。要做风险调整后才能判断 Alpha。

### 题 2

PB 越低越便宜，所以直接用 PB 作为价值因子，值越大越好，对吗？

答案：不对。PB 越低通常越偏价值，如果直接使用 PB，那么数值越大反而越贵。常见处理是使用 `-PB`、`1/PB`、`BP`，或 `-zscore(log(PB))`。

### 题 3

Alpha 是不是永远代表基金经理能力？

答案：不是。Alpha 是相对于某个模型未被解释的部分。如果模型遗漏了规模、价值、行业、动量等因素，所谓 Alpha 可能只是遗漏因子暴露。

### 题 4

为什么 Fama-French 模型重要？

答案：它展示了市场 Beta 之外，规模和价值等系统性风格也能解释资产收益差异。它让人们从“单纯选股能力”转向“收益来源拆解”。

### 题 5

一个好因子至少需要哪三点？

答案：可解释、可度量、可验证。

## D. 完整 Python 实验

下面这段代码是一个自包含实验。你可以单独复制到 Notebook 里运行。它的目的不是模拟真实市场，而是把今天主题的计算口径、方向、检查方法串起来。


```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(101)
dates = pd.bdate_range("2024-01-02", periods=180)
market = pd.Series(rng.normal(0.0004, 0.01, len(dates)), index=dates, name="market")
stock = 0.0002 + 1.25 * market + rng.normal(0, 0.012, len(dates))
stock = pd.Series(stock, index=dates, name="stock")

X = np.column_stack([np.ones(len(market)), market.values])
alpha, beta = np.linalg.lstsq(X, stock.values, rcond=None)[0]
residual = stock - (alpha + beta * market)

result = pd.Series({
    "daily_alpha": alpha,
    "beta": beta,
    "annualized_alpha_approx": alpha * 252,
    "residual_vol": residual.std() * np.sqrt(252),
})
print(result.round(4))
```
