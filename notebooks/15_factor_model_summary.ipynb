## 1. 因子投资的演进

### 1.1 什么是因子？

因子（Factor）是**能够解释资产收益差异的系统性特征**。

简单来说，因子就是**能够预测股票收益的变量**。

### 1.2 因子模型的演进

| 模型 | 年份 | 因子数量 | 核心思想 |
|------|------|----------|----------|
| **CAPM** | 1964 | 1 | 市场风险是唯一的风险溢价来源 |
| **三因子** | 1993 | 3 | 规模和价值也能解释收益差异 |
| **五因子** | 2015 | 5 | 盈利和投资也是重要因子 |

### 1.3 为什么要学习因子模型？

1. **理解收益来源**：知道赚的是什么钱
2. **风险归因**：知道承担了什么风险
3. **组合构建**：基于因子构建投资策略
4. **绩效评估**：评估基金经理的真正能力

## 2. CAPM：单因子模型

### 2.1 核心思想

**CAPM（Capital Asset Pricing Model）** 认为：

- 市场风险是唯一的风险溢价来源
- 资产的预期收益只取决于其 Beta（系统性风险）
- Alpha 应该为 0（如果没有选股能力）

### 2.2 公式

$$R_i - R_f = \beta_i (R_m - R_f) + \epsilon_i$$

其中：
- $R_i$：资产 i 的收益率
- $R_f$：无风险利率
- $R_m$：市场组合收益率
- $\beta_i$：资产 i 的 Beta（系统性风险）
- $\epsilon_i$：残差（非系统性风险）

### 2.3 CAPM 的局限

实证研究发现，CAPM 无法解释以下异象：

- **小市值效应**：小公司股票长期收益高于大公司
- **价值效应**：高 B/M 比（价值股）收益高于低 B/M 比（成长股）
- **动量效应**：过去表现好的股票未来表现也好

这些异象促使学者发展多因子模型。

---
## 3. Fama-French 三因子模型

### 3.1 核心思想

**Fama-French 三因子模型**（1993）认为：

- 除了市场风险，**规模**和**价值**也是风险溢价来源
- 小市值公司有更高的系统性风险
- 价值股有更高的系统性风险

### 3.2 公式

$$R_i - R_f = \alpha_i + \beta_i(R_m - R_f) + s_i \cdot SMB + h_i \cdot HML + \epsilon_i$$

新增的两个因子：

| 因子 | 全称 | 含义 | 构建方法 |
|------|------|------|----------|
| $SMB$ | Small Minus Big | 规模因子 | 小市值组合 - 大市值组合 |
| $HML$ | High Minus Low | 价值因子 | 高 B/M 组合 - 低 B/M 组合 |

### 3.3 因子构建方法

**SMB 构建**：
1. 每年 6 月底，按市值中位数将所有股票分为 Small (S) 和 Big (B) 两组
2. 计算每组的市值加权收益率
3. $SMB = R_S - R_B$

**HML 构建**：
1. 每年 6 月底，按 B/M 比的 30% 和 70% 分位数将股票分为三组
2. **High (H)**：B/M > 70% 分位数（价值股）
3. **Low (L)**：B/M < 30% 分位数（成长股）
4. $HML = R_H - R_L$

---
## 4. Fama-French 五因子模型

### 4.1 核心思想

**Fama-French 五因子模型**（2015）认为：

- 除了市场、规模、价值，**盈利**和**投资**也是风险溢价来源
- 高盈利公司有更强的抗风险能力
- 保守投资的公司现金流更稳定

### 4.2 公式

$$R_i - R_f = \alpha_i + \beta_i(R_m - R_f) + s_i \cdot SMB + h_i \cdot HML + r_i \cdot RMW + c_i \cdot CMA + \epsilon_i$$

新增的两个因子：

| 因子 | 全称 | 含义 | 构建方法 |
|------|------|------|----------|
| $RMW$ | Robust Minus Weak | 盈利因子 | 高盈利组合 - 低盈利组合 |
| $CMA$ | Conservative Minus Aggressive | 投资因子 | 保守投资组合 - 激进投资组合 |

### 4.3 因子构建方法

**RMW 构建**：
1. 每期按盈利能力（ROE）的 30%/70% 分位数将股票分为三组
2. **Robust (R)**：盈利能力 ≥ 70% 分位数（高盈利公司）
3. **Weak (W)**：盈利能力 ≤ 30% 分位数（低盈利公司）
4. $RMW = R_{robust} - R_{weak}$

**CMA 构建**：
1. 每期按投资率（总资产增长率）的 30%/70% 分位数将股票分为三组
2. **Conservative (C)**：投资率 ≤ 30% 分位数（保守投资公司）
3. **Aggressive (A)**：投资率 ≥ 70% 分位数（激进投资公司）
4. $CMA = R_{conservative} - R_{aggressive}$

## 5. 因子模型对比

### 5.1 模型对比表

| 模型 | 因子 | 优点 | 缺点 |
|------|------|------|------|
| **CAPM** | 市场 | 简单、理论基础扎实 | 无法解释规模和价值异象 |
| **三因子** | 市场+规模+价值 | 能解释更多异象 | 无法解释盈利和投资异象 |
| **五因子** | 市场+规模+价值+盈利+投资 | 解释力最强 | 因子多、可能过拟合 |

### 5.2 因子的经济含义

| 因子 | 经济含义 | 投资策略 |
|------|----------|----------|
| **市场 (Rm-Rf)** | 承担市场风险的补偿 | 被动指数投资 |
| **规模 (SMB)** | 小公司风险更高 | 小盘股策略 |
| **价值 (HML)** | 价值股风险更高 | 价值投资策略 |
| **盈利 (RMW)** | 高盈利公司更抗跌 | 质量投资策略 |
| **投资 (CMA)** | 保守投资公司更稳定 | 低资本支出策略 |

### 5.3 因子之间的关系

- **HML 和 RMW 负相关**：价值股通常盈利较低，高盈利股通常估值较高
- **HML 和 CMA 正相关**：价值股通常投资保守，成长股通常投资激进
- **SMB 和 CMA 负相关**：小公司通常投资激进，大公司通常投资保守

````python
# 模拟因子数据，展示因子之间的关系
np.random.seed(42)
n_periods = 60

# 生成因子收益
market_returns = np.random.normal(0.008, 0.05, n_periods)
smb_returns = np.random.normal(0.005, 0.03, n_periods)
hml_returns = np.random.normal(0.003, 0.025, n_periods)
rmw_returns = np.random.normal(0.004, 0.02, n_periods)
cma_returns = np.random.normal(0.003, 0.02, n_periods)

# 创建因子 DataFrame
factors_df = pd.DataFrame({
    'Market': market_returns,
    'SMB': smb_returns,
    'HML': hml_returns,
    'RMW': rmw_returns,
    'CMA': cma_returns
})

# 计算相关系数
corr_matrix = factors_df.corr()

# 绘制相关系数热力图
fig, ax = plt.subplots(figsize=(8, 6))
im = ax.imshow(corr_matrix, cmap='RdYlBu_r', vmin=-1, vmax=1)
ax.set_xticks(range(5))
ax.set_yticks(range(5))
ax.set_xticklabels(corr_matrix.columns, fontsize=10)
ax.set_yticklabels(corr_matrix.columns, fontsize=10)
ax.set_title('五因子相关系数矩阵', fontsize=14)

# 添加数值标签
for i in range(5):
    for j in range(5):
        ax.text(j, i, f'{corr_matrix.iloc[i, j]:.2f}',
                ha='center', va='center', fontsize=10,
                color='white' if abs(corr_matrix.iloc[i, j]) > 0.5 else 'black')

plt.colorbar(im, ax=ax)
plt.tight_layout()
plt.show()

print("\n解读：")
print("  - Market 与其他因子相关性较低（分散化效果好）")
print("  - HML 和 RMW 有较强的负相关（价值股 vs 高盈利股）")
print("  - SMB 和 CMA 有较强的负相关（小公司 vs 保守投资）")
````

## 6. 因子投资的实际应用

### 6.1 因子选股

基于因子模型选股的逻辑：

1. **计算每只股票的因子暴露**（通过回归）
2. **选择因子暴露符合预期的股票**
3. **构建投资组合**

例如，如果你想构建一个**小盘价值股**组合：
- 选择 SMB 暴露为正的股票（小市值）
- 选择 HML 暴露为正的股票（高 B/M）

### 6.2 因子择时

因子择时是指**根据市场环境调整因子暴露**：

| 市场环境 | 推荐因子 | 原因 |
|----------|----------|------|
| 经济扩张 | 规模 (SMB)、价值 (HML) | 小盘股和价值股在经济扩张期表现好 |
| 经济收缩 | 盈利 (RMW)、投资 (CMA) | 高盈利和保守投资公司在经济收缩期更抗跌 |
| 高波动 | 盈利 (RMW) | 高盈利公司波动更小 |

### 6.3 因子组合构建

**多因子组合**的构建方法：

1. **等权组合**：每个因子等权重
2. **风险平价**：每个因子贡献相同的风险
3. **优化组合**：最大化夏普比

```python
# 多因子选股示例
def multi_factor_select(stock_info, factors=['roe', 'pe_ratio', 'market_cap']):
    """
    多因子选股
    """
    # 计算每个因子的得分
    for factor in factors:
        stock_info[f'{factor}_score'] = rankdata(stock_info[factor]) / len(stock_info)
    
    # 综合得分
    stock_info['total_score'] = stock_info[[f'{f}_score' for f in factors]].mean(axis=1)
    
    # 选择得分最高的股票
    selected = stock_info.nlargest(10, 'total_score')
    return selected
```

## 7. 小结

### 核心收获

1. **因子投资的演进**：CAPM → 三因子 → 五因子

2. **五个核心因子**：
   - **市场 (Rm-Rf)**：承担市场风险的补偿
   - **规模 (SMB)**：小公司风险更高
   - **价值 (HML)**：价值股风险更高
   - **盈利 (RMW)**：高盈利公司更抗跌
   - **投资 (CMA)**：保守投资公司更稳定

3. **因子构建方法**：
   - 按分位数分组
   - 计算做多-做空组合的收益
   - 市值加权

4. **因子投资应用**：
   - 因子选股：选择因子暴露符合预期的股票
   - 因子择时：根据市场环境调整因子暴露
   - 因子组合：多因子组合构建

### 关键公式

- **CAPM**：$R_i - R_f = \beta_i (R_m - R_f)$
- **三因子**：$R_i - R_f = \alpha + \beta(R_m-R_f) + s \cdot SMB + h \cdot HML$
- **五因子**：$R_i - R_f = \alpha + \beta(R_m-R_f) + s \cdot SMB + h \cdot HML + r \cdot RMW + c \cdot CMA$
